# 06 - Drill a simple hole

We'll build a box and then drill a circular hole through it. The hole
is placed by sketching a single point on a plane and feeding that
sketch to `AddSimpleHole`.

**Prereq:** open a fresh empty part in Alibre.

## Setup: a 4 × 3 × 2 box

```python
from alibrex import (
    CurrentPart,
    ADDirectionType,
    ADHoleDepthCondition,
    ADPartFeatureEndCondition,
)

part = CurrentPart()

xy = part.DesignPlanes.Item(0)
sk = part.Sketches.AddSketch(None, xy, "Base")
figs = sk.Figures
figs.AddLine(0.0, 0.0, 4.0, 0.0)
figs.AddLine(4.0, 0.0, 4.0, 3.0)
figs.AddLine(4.0, 3.0, 0.0, 3.0)
figs.AddLine(0.0, 3.0, 0.0, 0.0)

part.Features.AddExtrudedBoss(
    sk, 2.0, ADPartFeatureEndCondition.AD_TO_DEPTH,
    None, None, 0.0,
    ADDirectionType.AD_ALONG_NORMAL, None, None, False,
    None, False,
    "Box", "Depth", "",
)
```

## Mark the hole location with a sketch point on XY

```python
hole_sk = part.Sketches.AddSketch(None, xy, "HoleCenter")
hole_sk.Figures.AddSketchPoint(2.0, 1.5)        # X, Y on the plane
```

## Drill through-all, 0.6 cm diameter

```python
hole = part.Features.AddSimpleHole(
    hole_sk,
    0.0,                                        # depth (ignored for through-all)
    0.6,                                        # diameter (cm)
    True,                                       # reversed - drill *into* the block
    None,                                       # tappedThread (none)
    ADHoleDepthCondition.AD_HOLE_THROUGH_ALL,
    None, None, 0.0,                            # to-geometry (unused)
    "Hole",
    "",                                         # depthParameterName (none)
)
hole.Name
```
