# Reading all properties: comprehensive walk

Two helpers in `alibrex` let you discover every readable property on a runtime object without leaving Python:

- `probe_object(obj, label)`: dumps every property + value + type
- `probe_collection(coll, label, limit=N)`: samples items from a `Count`/`Item(i)` collection

Both are read-only and never mutate the model.

**Prereq:** open a fresh empty part in Alibre.

## Setup: a small part with one feature

```python
from alibrex import (
    CurrentPart,
    connect,
    ADDirectionType,
    ADPartFeatureEndCondition,
    probe_object,
    probe_collection,
)

part = CurrentPart()
xy = part.DesignPlanes.Item(0)
sk = part.Sketches.AddSketch(None, xy, "Base")
figs = sk.Figures
figs.AddLine(0.0, 0.0, 4.0, 0.0)
figs.AddLine(4.0, 0.0, 4.0, 3.0)
figs.AddLine(4.0, 3.0, 0.0, 3.0)
figs.AddLine(0.0, 3.0, 0.0, 0.0)
part.Features.AddExtrudedBoss(
    sk, 2.0, ADPartFeatureEndCondition.AD_TO_DEPTH,
    None, None, 0.0,
    ADDirectionType.AD_ALONG_NORMAL, None, None, False,
    None, False,
    "Block", "Depth", "",
)
```

## Probe the root (`IADRoot`)

```python
root = connect()
probe_object(root, "root")
```

## Probe the active part (`IADPartSession`)

```python
probe_object(part, "part")
```

## Probe the first body (`IADBody`)

```python
body = part.Bodies.Item(0)
probe_object(body, "body")
```

## Probe a single face

```python
face = body.Faces.Item(0)
probe_object(face, "face[0]")
```

## Probe a single edge

```python
edge = body.Edges.Item(0)
probe_object(edge, "edge[0]")
```

## Probe a single vertex

```python
vertex = body.Vertices.Item(0)
probe_object(vertex, "vertex[0]")
```

## Probe a feature

```python
feat = part.Features.Item(0)
probe_object(feat, "feat[0]")
```

## Probe a parameter

```python
p = part.Parameters.Item(0)
probe_object(p, "param[0]")
```

## Probe the configurations container

```python
probe_object(part.Configurations, "Configurations")
```

## Sample collections with `probe_collection`

Walks `Count`/`Item(i)` collections and probes up to `limit` items.

```python
probe_collection(part.Parameters, "Parameters", limit=3)
```

```python
probe_collection(part.Features, "Features", limit=3)
```

```python
probe_collection(body.Faces, "Faces", limit=2)
```

```python
probe_collection(body.Edges, "Edges", limit=2)
```

## Free-form introspection

When `probe_*` is overkill, fall back to plain Python:

```python
# Every public name on the object
[n for n in dir(part) if not n.startswith('_')][:20]
```

```python
# The CLR class chain
type(body).__name__
```

```python
# narrow proxies when you want interface-specific autocomplete
from alibrex import IADBody, IADFace, IADEdge, narrow
body = narrow(body, IADBody)
face = narrow(face, IADFace)
edge = narrow(edge, IADEdge)
type(body).__name__, type(face).__name__, type(edge).__name__
```
