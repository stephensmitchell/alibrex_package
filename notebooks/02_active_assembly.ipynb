# 02 - The active assembly

Same idea as 01, but for assemblies. `CurrentAssembly()` errors if the
active document isn't an assembly.

**Prereq:** open any assembly in Alibre (File - New - Assembly).

```python
from alibrex import CurrentAssembly
```

```python
asm = CurrentAssembly()
asm.Name
```

How many top-level occurrences (parts and sub-assemblies)?

```python
asm.RootOccurrence.Occurrences.Count
```

List them:

```python
root = asm.RootOccurrence
for i in range(root.Occurrences.Count):
    occ = root.Occurrences.Item(i)
    print(f"  {i}: {occ.Name}")
```
