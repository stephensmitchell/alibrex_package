# 05 — Revolve a profile

Sketch a 2D outline, then revolve it around an axis to create a
rotational solid.

>  **Gotcha:** `AddRevolvedBoss` takes the angle in **radians**,
> not degrees. Wrap your angle in `math.radians(...)`.

**Prereq:** open a fresh empty part in Alibre.

## Attach + imports

```python
import math
from alibrex import CurrentPart

part = CurrentPart()
```

## Sketch a half-profile on XY

We'll revolve a rectangle around the Y axis to make a cylinder. The
profile lies in the XY plane; the bottom edge runs along Y=0.

```python
radius = 1.5
height = 4.0

xy = part.DesignPlanes.Item(0)
sk = part.Sketches.AddSketch(None, xy, "Profile")
figs = sk.Figures
figs.AddLine(0.0,    0.0,    radius, 0.0)
figs.AddLine(radius, 0.0,    radius, height)
figs.AddLine(radius, height, 0.0,    height)
figs.AddLine(0.0,    height, 0.0,    0.0)
```

## Revolve 360° around the Y axis

```python
y_axis = part.DesignAxes.Item(1)

feat = part.Features.AddRevolvedBoss(
    sk,
    None,                       # axis occurrence (not used for part-local axes)
    y_axis,                     # the axis to revolve around
    math.radians(360.0),        # full revolution, in radians!
    "Cylinder",                 # feature name
)
feat.Name
```
