# 3D sketch creation - every Add* method

3D sketches let you place points, lines, and arcs in full XYZ space
(not constrained to a plane). They live under `part.Sketches3D`, not
`part.Sketches`.

**Prereq:** open a fresh empty part in Alibre.

## Setup - create a 3D sketch

```python
from alibrex import CurrentPart

part = CurrentPart()
sk3 = part.Sketches3D.Add3DSketch("My3DSketch")
sk3.Name
```

## Points - `AddPoint(x, y, z)` - `IAD3DSketchPoint`

```python
p1 = sk3.Figures.AddPoint(0.0, 0.0, 0.0)
p2 = sk3.Figures.AddPoint(2.0, 1.0, 3.0)
print(f"p1 = ({p1.X}, {p1.Y}, {p1.Z})")
print(f"p2 = ({p2.X}, {p2.Y}, {p2.Z})")
```

## Lines - `AddLine(x1, y1, z1, x2, y2, z2)` - `IAD3DSketchLine`

Two overloads: six floats, or two `IADPoint` arguments.

```python
line = sk3.Figures.AddLine(0.0, 0.0, 0.0, 5.0, 3.0, 2.0)
print(f"start = ({line.Start.X}, {line.Start.Y}, {line.Start.Z})")
print(f"end   = ({line.End.X}, {line.End.Y}, {line.End.Z})")
print(f"length = {line.Length:.4f}")
```

## Circular arc - center + start + end points (3D)

`AddCircularArcByCenterStartEnd(cx, cy, cz, sx, sy, sz, ex, ey, ez)` -
`IAD3DSketchCircularArc`. Center and both endpoints must be coplanar.

```python
arc = sk3.Figures.AddCircularArcByCenterStartEnd(
    0.0, 0.0, 0.0,         # center
    2.0, 0.0, 0.0,         # start
    0.0, 2.0, 0.0,         # end (quarter arc in XY)
)
```

## B-spline by interpolation

`AddBsplineByInterpolation(points)` builds a smooth curve through the
given XYZ control points. Use `float_array` from alibrex to feed the
CLR array. By-ref output args are surfaced as a tuple - the spline is
the first element.

```python
from alibrex import float_array

pts = float_array([0.0, 0.0, 0.0,
                   1.0, 2.0, 1.0,
                   3.0, 1.0, 2.0,
                   5.0, 3.0, 1.5])
spline, _ = sk3.Figures.AddBsplineByInterpolation(pts)
print(f"spline start = ({spline.StartPoint.X}, {spline.StartPoint.Y}, {spline.StartPoint.Z})")
print(f"spline end   = ({spline.EndPoint.X}, {spline.EndPoint.Y}, {spline.EndPoint.Z})")
```

## Summary table

| Method | Return type |
|---|---|
| `AddPoint(x, y, z)` | `IAD3DSketchPoint` |
| `AddPoint(IADPoint)` | `IAD3DSketchPoint` |
| `AddLine(x1, y1, z1, x2, y2, z2)` | `IAD3DSketchLine` |
| `AddLine(start_pt, end_pt)` | `IAD3DSketchLine` |
| `AddCircularArcByCenterStartEnd(...)` | `IAD3DSketchCircularArc` |
| `AddBsplineByInterpolation(points)` | `tuple[IAD3DSketchBspline, ...]` |
| `AddBspline(order, count, ctrl, knots, weights)` | `tuple[IAD3DSketchBspline, ...]` |
| `AddPolyline(points)` | `tuple[IObjectCollector, ...]` |
