# 09 - Walk an assembly tree

Assemblies form a tree: top-level occurrences may themselves be
sub-assemblies containing more occurrences. This notebook walks the
tree two levels deep using only `CurrentAssembly()`.

**Prereq:** open an existing assembly in Alibre - any saved assembly
file works.

## Attach

```python
from alibrex import CurrentAssembly

asm = CurrentAssembly()
asm.Name
```

## Top-level occurrence count

```python
asm.RootOccurrence.Occurrences.Count
```

## Names of top-level occurrences

```python
root = asm.RootOccurrence
for i in range(root.Occurrences.Count):
    occ = root.Occurrences.Item(i)
    print(f"  {occ.Name}")
```

## Walk one level deeper

For each top-level occurrence, list its sub-occurrences too (a
sub-assembly will have some; a leaf part won't).

```python
root = asm.RootOccurrence
for i in range(root.Occurrences.Count):
    occ = root.Occurrences.Item(i)
    print(f"  {occ.Name}")
    for j in range(occ.Occurrences.Count):
        sub = occ.Occurrences.Item(j)
        print(f"    └─ {sub.Name}")
```

## How many leaves are in the tree?

```python
root = asm.RootOccurrence
leaves = 0
branches = 0
for i in range(root.Occurrences.Count):
    occ = root.Occurrences.Item(i)
    if occ.Occurrences.Count == 0:
        leaves += 1
    else:
        branches += 1
        leaves += occ.Occurrences.Count
leaves, branches
```
