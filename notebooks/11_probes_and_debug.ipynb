# 11 - Probing and debugging

The AlibreX API is huge. When you've got a runtime object and don't
know what it can do, two helpers in `alibrex` let you discover its
shape without leaving Python:

- `probe_object(obj, label)` - print every readable property
- `probe_collection(coll, label, limit=N)` - sample items from a
  collection like `Bodies`, `Faces`, `Parameters`

Both are read-only; they never mutate the model.

**Prereq:** open a fresh empty part in Alibre.

## Setup: a small part with one feature

```python
from alibrex import (
    CurrentPart,
    ADDirectionType,
    ADPartFeatureEndCondition,
    probe_object,
    probe_collection,
)

part = CurrentPart()

xy = part.DesignPlanes.Item(0)
sk = part.Sketches.AddSketch(None, xy, "Base")
figs = sk.Figures
figs.AddLine(0.0, 0.0, 3.0, 0.0)
figs.AddLine(3.0, 0.0, 3.0, 2.0)
figs.AddLine(3.0, 2.0, 0.0, 2.0)
figs.AddLine(0.0, 2.0, 0.0, 0.0)
part.Features.AddExtrudedBoss(
    sk, 1.0, ADPartFeatureEndCondition.AD_TO_DEPTH,
    None, None, 0.0,
    ADDirectionType.AD_ALONG_NORMAL, None, None, False,
    None, False,
    "Box", "Depth", "",
)
```

## Probe the part - every readable property at a glance

```python
probe_object(part, "active part")
```

## Probe its Parameters collection (sample first 3)

```python
probe_collection(part.Parameters, "Parameters", limit=3)
```

## Probe a single face - useful for face/edge work

```python
body = part.Bodies.Item(0)
face = body.Faces.Item(0)
probe_object(face, "face[0]")
```

## Free-form: just use Python

`dir(obj)` lists names; `type(obj).__name__` shows the Python proxy class;
use `narrow(obj, IADBody)` when you want interface-specific autocomplete.

```python
from alibrex import IADBody, narrow
typed_body = narrow(body, IADBody)
type(typed_body).__name__, typed_body.Faces.Count
```
