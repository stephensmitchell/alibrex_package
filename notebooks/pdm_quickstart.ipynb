# PDM quickstart - connect, list safes, inspect a safe

Alibre's Product Data Management (PDM) layer is reachable through the
same `alibrex` runtime as parts/assemblies - there's an `IADRoot.GetActiveServerConnection()`
for reusing an existing UI login, and `IADRoot.ConnectToPDM(url, domain, user, password)`
for direct connections.

**Prereq:** either log in to PDM via Alibre's PDM panel (then this
notebook reuses that connection), or edit the four constants below
with your PDM server URL and credentials.

## Setup

```python
from alibrex import connect

root = connect()
print(f"Alibre version: {root.Version}")
```

## Get a PDM connection

Try the active UI connection first; fall back to direct connect.
Edit the four constants below if no UI connection is active.

```python
# === EDIT IF NEEDED =====================================
PDM_URL      = "http://localhost:8099/"
PDM_DOMAIN   = ""
PDM_USER     = ""     # your PDM username
PDM_PASSWORD = ""     # your PDM password
# ========================================================

try:
    conn = root.GetActiveServerConnection()
    if conn is None:
        raise RuntimeError("no active session")
except Exception:
    print("  No active connection - connecting directly ...")
    try:
        conn = root.ConnectToPDM(PDM_URL, PDM_DOMAIN, PDM_USER, PDM_PASSWORD)
    except Exception as exc:
        print(f"  ConnectToPDM failed: {exc}")
        raise SystemExit(1)

print(f"Server : {conn.URL}")
print(f"Domain : {conn.Domain}")
print(f"User   : {conn.UserName}")
print(f"Online : {conn.IsOnline}")
```

## List safes

```python
safes = conn.Safes
print(f"Found {safes.Count} safe(s)")
for i in range(safes.Count):
    s = safes.Item(i)
    print(f"  [{i}] {s.Name!r}  (ref: {s.Reference})")

safe = safes.Item(0)
safe.Name
```

## Property definitions on the safe

Property definitions are the metadata fields that file items can
carry. Built-in definitions ship with PDM; custom ones are added by
an admin.

```python
defs = safe.PropertyDefinitions
print(f"Definitions: {defs.Count}")
for i in range(defs.Count):
    pd = defs.Item(i)
    flavor = "built-in" if pd.IsBuiltInPropertyDefinition else "custom"
    print(f"  {pd.DisplayName!r:30s}  {pd.PropertyValueType}  {flavor}")
```

## Classes

Classes carry typed data items and default property values. The
PDM templates assemble classes into multi-level workflows.

```python
classes = safe.Classes
print(f"Classes: {classes.Count}")
for i in range(classes.Count):
    cls = classes.Item(i)
    kind = "core" if cls.IsCoreClass else "custom"
    print(f"  {cls.Name!r:30s}  {kind}  items={cls.DataItems.Count}")
```

## Templates

```python
templates = safe.Templates
print(f"Templates: {templates.Count}")
for i in range(templates.Count):
    t = templates.Item(i)
    print(f"  {t.Name!r}  levels={t.Levels.Count}")
```
