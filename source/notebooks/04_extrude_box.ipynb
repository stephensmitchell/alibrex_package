# 04: Extrude a box

Builds on 03. Sketch a rectangle, then extrude it into a 3D solid box. This is the canonical part-modeling pipeline:

    plane  →  sketch  →  figures  →  feature

**Prereq:** open a fresh empty part in Alibre.

## Attach + imports

```python
from alibrex import CurrentPart, ADPartFeatureEndCondition, ADDirectionType

part = CurrentPart()
```

## Sketch the base rectangle

```python
xy = part.DesignPlanes.Item(0)
sk = part.Sketches.AddSketch(None, xy, "Base")
figs = sk.Figures
figs.AddLine(0.0, 0.0, 5.0, 0.0)
figs.AddLine(5.0, 0.0, 5.0, 3.0)
figs.AddLine(5.0, 3.0, 0.0, 3.0)
figs.AddLine(0.0, 3.0, 0.0, 0.0)
```

## Extrude 1 cm along the sketch normal

```python
feat = part.Features.AddExtrudedBoss(
    sk,
    1.0,                                       # depth (cm)
    ADPartFeatureEndCondition.AD_TO_DEPTH,
    None, None, 0.0,                           # to-geometry (unused for blind depth)
    ADDirectionType.AD_ALONG_NORMAL,           # extrude along the sketch normal
    None, None, False,                         # direction overrides
    None, False,                               # draft
    "Box",                                     # feature name
    "Depth", "",                               # parameter names
)
feat.Name
```
