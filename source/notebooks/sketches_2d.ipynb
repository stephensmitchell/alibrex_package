# 2D sketch creation: every Add* method

Comprehensive reference for `IADSketchFigures.Add*`. Each section creates a fresh sketch on the XY plane and adds one figure type, so you can see each output in Alibre side by side.

**Prereq:** open a fresh empty part in Alibre.

## Setup

```python
import math
from alibrex import CurrentPart

part = CurrentPart()
xy = part.DesignPlanes.Item(0)
```

## Points: `AddSketchPoint(x, y)` returns `IADSketchPoint`

```python
sk = part.Sketches.AddSketch(None, xy, "Points")
pt1 = sk.Figures.AddSketchPoint(0.0, 0.0)
pt2 = sk.Figures.AddSketchPoint(2.0, 1.0)
print(f"pt1 = ({pt1.X}, {pt1.Y})")
print(f"pt2 = ({pt2.X}, {pt2.Y})")
```

## Lines: `AddLine(x1, y1, x2, y2)` returns `IADSketchLine`

```python
sk = part.Sketches.AddSketch(None, xy, "Lines")
line = sk.Figures.AddLine(0.0, 0.0, 3.0, 2.0)
print(f"start  = ({line.Start.X}, {line.Start.Y})")
print(f"end    = ({line.End.X}, {line.End.Y})")
print(f"length = {line.Length:.4f}")
```

## Circles: `AddCircle(cx, cy, r)` returns `IADSketchCircle`

```python
sk = part.Sketches.AddSketch(None, xy, "Circles")
c = sk.Figures.AddCircle(0.0, 0.0, 2.0)
print(f"center = ({c.Center.X}, {c.Center.Y})")
print(f"radius = {c.Radius}")
```

## Circular arc: center + start point + sweep angle

`AddCircularArc(cx, cy, sx, sy, angle_rad)` returns `IADSketchCircularArc`. Angle is **radians**, sweep direction CCW.

```python
sk = part.Sketches.AddSketch(None, xy, "Arc_Sweep")
arc = sk.Figures.AddCircularArc(0.0, 0.0, 2.0, 0.0, math.radians(90.0))
print(f"center = ({arc.Center.X}, {arc.Center.Y})  r={arc.Radius}")
```

## Circular arc: center + start + end points

```python
sk = part.Sketches.AddSketch(None, xy, "Arc_CSE")
arc = sk.Figures.AddCircularArcByCenterStartEnd(0.0, 0.0, 2.0, 0.0, 0.0, 2.0)
```

## Circular arc: by 3 points (start, midpoint, end)

```python
sk = part.Sketches.AddSketch(None, xy, "Arc_3pt")
arc = sk.Figures.AddCircularArcBy3Points(2.0, 0.0, 1.4142, 1.4142, 0.0, 2.0)
```

## Ellipses: `AddEllipse(cx, cy, major, minor/major_ratio, angle_rad)`

- returns `IADSketchEllipse`

```python
sk = part.Sketches.AddSketch(None, xy, "Ellipses")
e = sk.Figures.AddEllipse(0.0, 0.0, 3.0, 0.5, 0.0)
```

## Elliptic arc

```python
sk = part.Sketches.AddSketch(None, xy, "EllipticArc")
ea = sk.Figures.AddEllipticArc(0.0, 0.0, 3.0, 0.5, 3.0, 0.0, 0.0, 1.5, 0.0)
```

## Rectangle: `AddRectangle(xlow, ylow, xhigh, yhigh)`

Returns an `IObjectCollector` of 4 edges instead of a single figure.

```python
sk = part.Sketches.AddSketch(None, xy, "Rectangle")
edges = sk.Figures.AddRectangle(0.0, 0.0, 5.0, 3.0)
print(f"Rectangle returned {edges.Count} edges")
```

## Summary table

| Method | Return type |
|---|---|
| `AddSketchPoint(x, y)` | `IADSketchPoint` |
| `AddLine(x1, y1, x2, y2)` | `IADSketchLine` |
| `AddCircle(cx, cy, r)` | `IADSketchCircle` |
| `AddCircularArc(cx, cy, sx, sy, angle)` | `IADSketchCircularArc` |
| `AddCircularArcByCenterStartAngle(...)` | `IADSketchCircularArc` |
| `AddCircularArcByCenterStartEnd(...)` | `IADSketchCircularArc` |
| `AddCircularArcBy3Points(...)` | `IADSketchCircularArc` |
| `AddEllipse(cx, cy, major, ratio, angle)` | `IADSketchEllipse` |
| `AddEllipseBy3Points(...)` | `IADSketchEllipse` |
| `AddEllipticArc(...)` | `IADSketchEllipticArc` |
| `AddRectangle(...)` | `IObjectCollector` (4 edges) |
| `AddBspline(...)` / `AddBsplineByInterpolation(...)` | `IADSketchBspline` (advanced: needs `float_array` for control points) |

Every `Add*` method returns its specific subclass. Save the return value to keep the precise type and avoid `Item(N)` re-fetches.
